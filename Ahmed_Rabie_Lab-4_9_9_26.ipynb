```python
!pip install -U google-genai
from google.colab import userdata
api_key = userdata.get('Gemini_API_Key')
client = genai.Client(api_key=api_key)
MODEL = "models/gemini-3.6-flash"

university_data = {
    "Campuses": {
        "Federal Urdu University Islamabad Campus",
        "Federal Urdu University Karachi Gulshan Campus",
        "Federal Urdu University Karachi Abdul Haq Campus",
    },
    "Federal Urdu University Karachi Gulshan Campus":{
    "Subjects_in_CS_6th_Semester": {
        "Artificial Intelligence",
        "Professional Practices",
        "Natural Language Processing",
        "Numerical and Symbolic Computing",
        "Compiler Construction",
    },
    "Subject_Teachers":{
        "Compiler Construction": "Miss Naheed",
        "Natural Language Processing": "Sir Sarim",
        "Artificial Intelligence": "Miss Uzma",
        "Numerical and Symbolic Computing": "Sir Akhter",
        "Professional Practices": "Not Assigned Yet",
    }
    }

}

def university_agent(user_input):

    prompt = f"""
You are a helpful university assistant chatbot for our university.
Use ONLY the data provided below to answer questions. Do not guess or make up
any information that is not present in this data.

All Campuses:
{university_data["Campuses"]}

Data below is ONLY available for Federal Urdu University Karachi Gulshan Campus:

BSCS 6th Semester Subjects:
{university_data["Federal Urdu University Karachi Gulshan Campus"]["Subjects_in_CS_6th_Semester"]}

Subject-Teacher mapping:
{university_data["Federal Urdu University Karachi Gulshan Campus"]["Subject_Teachers"]}

Rules:
1. You only have detailed subject and teacher data for Federal Urdu University
   Karachi Gulshan Campus. You do NOT have subject or teacher data for any
   other campus.
2. If the student asks about subjects, teachers, or any other detail about any
   other campus, politely explain that you don't have that information
   available.
5. Keep responses short, clear, and easy for students to read.

User request: {user_input}
Response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text

print("Gemini University Agent Started (type 'exit' to stop)\n")
while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = university_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
